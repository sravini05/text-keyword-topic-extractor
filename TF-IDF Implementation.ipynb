```python
import pandas as pd

msg = pd.read_csv(
    "C:/Users/prath/Downloads/spam.csv",
 
    encoding="ISO-8859-1"   # Or encoding="latin1"
)

print(msg.head())
```

```python
msg = msg.drop(columns=["Unnamed: 2" ,'Unnamed: 3',"Unnamed: 4"],axis=1)
```

```python
msg
```

```python
import nltk
import re
```

```python
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
corpus = []                     ## we use the lemmatization 
for i in range(0,len(msg)):
    review = re.sub('[^a-zA-Z]',' ' ,msg['v2'][i])
    review = review.lower()
    review = review.split()
    review =[lemmatizer.lemmatize(word.lower()) for word in review if not word.lower() in stopwords.words('english')]
    review = ' '.join(review)
    corpus.append(review)
```

### Create TF-IDF And N_grams

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(max_features=100)
x = tfidf.fit_transform(corpus).toarray()
```

```python
import numpy as np
np.set_printoptions(edgeitems = 30 , linewidth=100000,
                    formatter=dict(float=lambda x:"%.3g"%x))
```
